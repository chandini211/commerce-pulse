- Reading all csv files(Datbases)

```python
import pandas as pd
from pathlib import Path

# Path to raw data folder
DATA_PATH = Path("../data/raw")

# Read all CSV files
customers = pd.read_csv(DATA_PATH / "customers.csv")
orders = pd.read_csv(DATA_PATH / "orders.csv")
order_items = pd.read_csv(DATA_PATH / "order_items.csv")
payments = pd.read_csv(DATA_PATH / "order_payments.csv")
reviews = pd.read_csv(DATA_PATH / "order_reviews.csv")
products = pd.read_csv(DATA_PATH / "products.csv")
sellers = pd.read_csv(DATA_PATH / "sellers.csv")
geolocation = pd.read_csv(DATA_PATH / "geolocation.csv")
category = pd.read_csv(DATA_PATH / "product_category_name_translation.csv")
```

- Storing in dictionary and knowing the sizes of each database

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order Items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "geolocation": geolocation,
    "category": category,
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

- Primary key Validation

```python
# Check uniqueness of potential primary keys

primary_keys = {
    "customers": "customer_id",
    "orders": "order_id",
    "products": "product_id",
    "sellers": "seller_id"
}

for table, pk in primary_keys.items():
    df = datasets[table]

    print("="*50)
    print(f"Table: {table}")
    print(f"Primary Key: {pk}")

    print("Total Rows :", len(df))
    print("Unique Keys:", df[pk].nunique())

    if len(df) == df[pk].nunique():
        print("Valid Primary Key")
    else:
        print("Duplicate Primary Keys Found")
```

```python
def validate_primary_key(df, table_name, primary_key):
    """
    Validates whether a column can act as a Primary Key.
    """

    total_rows = len(df)
    unique_values = df[primary_key].nunique()

    print("=" * 60)
    print(f"Table Name   : {table_name}")
    print(f"Primary Key  : {primary_key}")
    print(f"Total Rows   : {total_rows}")
    print(f"Unique Values: {unique_values}")

    if total_rows == unique_values:
        print("Primary Key is VALID")
    else:
        print("Primary Key is INVALID")
```

```python
validate_primary_key(customers, "Customers", "customer_id")
validate_primary_key(orders, "Orders", "order_id")
validate_primary_key(products, "Products", "product_id")
validate_primary_key(sellers, "Sellers", "seller_id")
```

- Understanding the quality and structure of data before transforming

```python
def profile_dataset(df, table_name):
    print("=" * 70)
    print(f"TABLE : {table_name.upper()}")
    print("=" * 70)

    print(f"\nShape : {df.shape}")

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nMemory Usage:")
    print(df.memory_usage(deep=True).sum() / 1024**2, "MB")
```

```python
for name, df in datasets.items():
    profile_dataset(df, name)
```

```python
def validate_foreign_key(parent_df, child_df, parent_key, child_key, parent_name, child_name):
    """
    Validates referential integrity between two tables.
    """
    
    orphan_records = child_df[~child_df[child_key].isin(parent_df[parent_key])]

    print("=" * 70)
    print(f"{child_name} --> {parent_name}")
    print("=" * 70)

    print(f"Child Records : {len(child_df)}")
    print(f"Orphan Records: {len(orphan_records)}")

    if orphan_records.empty:
        print("Referential Integrity PASSED")
    else:
        print("Referential Integrity FAILED")
        print(orphan_records.head())
```

```python
validate_foreign_key(customers, orders,
                     "customer_id", "customer_id",
                     "Customers", "Orders")

validate_foreign_key(orders, order_items,
                     "order_id", "order_id",
                     "Orders", "Order Items")

validate_foreign_key(products, order_items,
                     "product_id", "product_id",
                     "Products", "Order Items")

validate_foreign_key(sellers, order_items,
                     "seller_id", "seller_id",
                     "Sellers", "Order Items")

validate_foreign_key(orders, payments,
                     "order_id", "order_id",
                     "Orders", "Payments")

validate_foreign_key(orders, reviews,
                     "order_id", "order_id",
                     "Orders", "Reviews")
```
